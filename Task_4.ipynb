```python
import pandas as pd
```

```python
df = pd.read_csv("/content/Indian FMCG Retail Sales  Customer  Inventory (2024).csv")
```

```python
print("Original rows:", len(df))
print("Original columns:", len(df.columns))
df = df.drop_duplicates()
df = df.drop_duplicates(subset=["Invoice_ID"])
```

```python
df["Invoice_Date"] = pd.to_datetime(df["Invoice_Date"], errors="coerce")
```

```python
numeric_columns = ["Invoice_ID","Units","Cost_Price","Selling_Price","Revenue","Cost","Margin","Margin_Percent","Stock_On_Hand","Reorder_Level","Lead_Time_Days","Customer_Age","Loyalty_Flag"]
```

```python
for column in numeric_columns:
    if column in df.columns:
        df[column] = pd.to_numeric(df[column], errors="coerce")
if "Customer_Age" in df.columns:
    df["Customer_Age"] = df["Customer_Age"].fillna(
        df["Customer_Age"].median()
    )
for column in numeric_columns:
    if column in df.columns:
        df[column] = df[column].fillna(0)
```

```python
integer_columns = ["Invoice_ID","Units","Stock_On_Hand","Reorder_Level","Lead_Time_Days","Customer_Age","Loyalty_Flag"]
```

```python
for column in integer_columns:
    if column in df.columns:
        df[column] = df[column].astype(int)
if "Invoice_ID" in df.columns:
    df = df[df["Invoice_ID"] > 0]
if "Units" in df.columns:
    df = df[df["Units"] >= 0]
if "Selling_Price" in df.columns:
    df = df[df["Selling_Price"] >= 0]
```

```python
df.to_csv("retail_cleaned.csv", index=False)
print("Cleaning completed successfully.")
print("Final rows:", len(df))
print("Final columns:", len(df.columns))
print("\nMissing values:")
print(df.isnull().sum())
```
