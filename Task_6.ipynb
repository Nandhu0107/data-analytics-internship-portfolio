```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/movie_ratings_analysis_650_movies.csv")
print(df.head())
```

```python
print(df.columns)
```

```python
genre_count = df["Genre"].value_counts()
```

```python
plt.figure(figsize=(10, 5))
genre_count.plot(kind="bar")
plt.title("Movie Genre Distribution")
plt.xlabel("Genre")
plt.ylabel("Number of Movies")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("genre_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Budget_Million"], df["Revenue_Million"])
plt.title("Budget vs Revenue")
plt.xlabel("Budget (Million)")
plt.ylabel("Revenue (Million)")
plt.tight_layout()
plt.savefig("budget_vs_revenue.png")
plt.show()
```

```python
country_count = df["Country"].value_counts()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    country_count,
    labels=country_count.index,
    autopct="%1.1f%%"
)
plt.title("Movies by Country")
plt.tight_layout()
plt.savefig("movies_by_country.png")
plt.show()
```

```python
language_count = df["Language"].value_counts()
```

```python
language_rating = df.groupby("Language")["Rating"].mean()
```

```python
plt.figure(figsize=(8, 8))
plt.pie(
    language_count,
    labels=language_count.index,
    autopct="%1.1f%%",
    pctdistance=0.85,
    wedgeprops=dict(width=0.4, edgecolor='w')
)
plt.title("Movies by Language")
plt.tight_layout()
plt.savefig("movies_by_language.png")
plt.show()
```

```python
plt.figure(figsize=(10, 5))
language_rating.plot(kind="line")
plt.title("Average Rating by Language")
plt.xlabel("Language")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("average_rating_language.png")
plt.show()
```

```python
genre_revenue = df.groupby("Genre")["Revenue_Million"].mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(10, 5))
genre_revenue.plot(kind="area")
plt.title("Average Revenue by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Revenue (Million)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("average_revenue_genre.png")
plt.show()
```
