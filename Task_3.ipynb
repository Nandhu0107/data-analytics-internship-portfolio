```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("/content/dirty_ecommerce_dataset.csv")
df.head()
```

```python
print("Before Cleaning")
print(df.head())
print("Shape:", df.shape)
print("Missing values:")
print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
before_rows = df.shape[0]
before_columns = df.shape[1]
before_missing = df.isnull().sum().sum()
before_duplicates = df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
```

```python
for col in df.select_dtypes(include="object"):
    df[col] = df[col].str.strip()
```

```python
for col in df.select_dtypes(include="number"):
    df[col] = df[col].fillna(df[col].median())
for col in df.select_dtypes(include="object"):
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
for col in df.columns:
    if "date" in col:
        df[col] = pd.to_datetime(df[col], errors="coerce")
df = df.dropna(how="all")
```

```python
after_rows = df.shape[0]
after_columns = df.shape[1]
after_missing = df.isnull().sum().sum()
after_duplicates = df.duplicated().sum()
```

```python
print("After Cleaning")
print(df.head())
print("Shape:", df.shape)
print("Missing values:")
print(df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python
comparison = pd.DataFrame({
    "Metric": ["Rows", "Columns", "Missing Values", "Duplicate Rows"],
    "Before": [before_rows, before_columns, before_missing, before_duplicates],
    "After": [after_rows, after_columns, after_missing, after_duplicates]
})
print("\nBefore and After Comparison")
display(comparison)
```

```python
df.to_csv("cleaned_dataset.csv", index=False)
comparison.to_csv("before_after_comparison.csv", index=False)
print("\nCleaned dataset saved successfully.")
```
