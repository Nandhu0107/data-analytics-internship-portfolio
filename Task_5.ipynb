```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sn
```

```python
df=pd.read_csv('/content/telecom_customer_churn_eda-1.csv')
df.head(10)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['ChurnReason'].value_counts()
```

```python
df['MonthlyCharges'] = df['MonthlyCharges'].fillna(df['MonthlyCharges'].median())
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())
df['ChurnReason'] = df['ChurnReason'].fillna('Not Applicable')
```

```python
df['ChurnReason'].value_counts()
```

```python
df.isnull().sum()
```

```python
print(df['Churn'].value_counts())
df['Churn'].value_counts(normalize=True) * 100
df[['tenure', 'MonthlyCharges', 'TotalCharges']].describe()
```

```python
sn.countplot(data=df, x='Churn')
plt.title('Customer Churn Distribution')
plt.show()
```

```python
sn.countplot(data=df, x='Contract', hue='Churn')
plt.title('Churn by Contract Type')
plt.xticks(rotation=20)
plt.show()
```

```python
sn.histplot(data=df, x='tenure', hue='Churn', kde=True)
plt.title('Tenure Distribution by Churn')
plt.show()
```

```python
sn.boxplot(data=df, x='Churn', y='MonthlyCharges')
plt.title('Monthly Charges vs Churn')
plt.show()
```

```python

sn.boxplot(data=df, x='Churn', y='TotalCharges')
plt.title('Total Charges vs Churn')
plt.show()
```

```python
sn.boxplot(x=df['MonthlyCharges'])
plt.title('Monthly Charges Outliers')
plt.show()
```

```python
sn.boxplot(x=df['TotalCharges'])
plt.title('Total Charges Outliers')
plt.show()
```

```python
Q1 = df['MonthlyCharges'].quantile(0.25)
Q3 = df['MonthlyCharges'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['MonthlyCharges']<lower)|(df['MonthlyCharges']>upper)]
print("Number of outliers:",len(outliers))
```

```python
Q1 = df['TotalCharges'].quantile(0.25)
Q3 = df['TotalCharges'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
total_outliers = df[(df['TotalCharges']<lower)|(df['TotalCharges'] > upper)]
print("TotalCharges outliers:", len(total_outliers))
```

```python
plt.figure(figsize=(10,6))
df['ChurnReason'].value_counts().plot(kind='bar')
plt.title('Customer Churn Reasons')
plt.xlabel('Churn Reason')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45,ha='right')
plt.tight_layout()
plt.show()
```

```python
numeric_cols=['SeniorCitizen','tenure','MonthlyCharges','TotalCharges']
corr=df[numeric_cols].corr()
plt.figure(figsize=(8,5))
sn.heatmap(corr,annot=True,cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()
```

```python
print("Churn Rate:")
print(df['Churn'].value_counts(normalize=True).mul(100).round(2))
```

```python
print("Churn by Contract:")
print(pd.crosstab(df['Contract'],df['Churn'],normalize='index').mul(100).round(2))
```

```python
print("Average Tenure by Churn:")
print(df.groupby('Churn')['tenure'].mean().round(2))
print("Average Monthly Charges by Churn:")
print(df.groupby('Churn')['MonthlyCharges'].mean().round(2))
print("Top Churn Reasons:")
print(df[df['Churn']=='Yes']['ChurnReason'].value_counts().head())
```
